## 1. Importar bibliotecas
**Concepto:** las bibliotecas proporcionan herramientas para trabajar con datos, dividirlos y evaluar modelos.

```python
import numpy as np
from sklearn import datasets
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_val_score, KFold, train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

## 2. Cargar el conjunto Iris
**Concepto:** Iris contiene medidas de flores y la especie a la que pertenece cada una.

```python
# Cargamos las características y sus clases.
iris = datasets.load_iris()
X = iris.data
y = iris.target
print("Clases:", iris.target_names)
```

## 3. Examinar los datos
**Concepto:** una instancia es un registro y un atributo es una característica medida.

```python
print("Tabla de datos: %d instancias y %d atributos" % (X.shape[0], X.shape[1]))
print("Valores de la clase:", set(y))
print("Atributos:", iris.feature_names)
```

## 4. Separar entrenamiento y prueba (80 % / 20 %)
**Concepto:** el conjunto de prueba se reserva para evaluar el modelo con datos no vistos. `stratify=y` mantiene la proporción de especies.

```python
# Partición externa: 80 % para entrenamiento, 20 % para prueba.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Registros de entrenamiento:", len(X_train))
print("Registros de prueba:", len(X_test))
```

## 5. Estandarización sin fuga de información
**Concepto:** `StandardScaler` centra las variables y ajusta su escala. Un `Pipeline` calcula la estandarización **solo con los datos de entrenamiento de cada partición**.

```python
# El Pipeline conecta la estandarización con el clasificador.
# Así, cada partición de la validación cruzada ajusta su propio escalador.
modelo = make_pipeline(StandardScaler(), SVC(C=0.5, gamma="auto"))
print("Pipeline creado:", modelo)
```

## 6. Validación cruzada con cinco particiones
**Concepto:** se divide el entrenamiento en cinco partes: en cada vuelta se valida con una parte y se entrena con las otras cuatro.

**Importante:** aquí se evalúan los parámetros fijos `C=0.5` y `gamma="auto"`; no se realiza una búsqueda de hiperparámetros.

```python
# shuffle=True mezcla los registros de forma reproducible.
cv = KFold(n_splits=5, shuffle=True, random_state=42)
resultados = cross_val_score(modelo, X_train, y_train, cv=cv, scoring="accuracy")
print("Exactitud por partición:", np.round(resultados, 4))
print("Accuracy (media +/- desv.): %0.4f +/- %0.4f" % (resultados.mean(), resultados.std()))
```

## 7. Entrenar el modelo definitivo
**Concepto:** después de la validación cruzada, el modelo se ajusta con **todo el conjunto de entrenamiento**, sin usar el de prueba.

```python
modelo.fit(X_train, y_train)
print("Modelo definitivo entrenado.")
```

## 8. Evaluar la exactitud en prueba
**Concepto:** la exactitud (*accuracy*) es la proporción de predicciones correctas. El conjunto de prueba solo se usa al final.

```python
test_results = modelo.score(X_test, y_test)
print("Exactitud en test: %.2f %%" % (test_results * 100))
```

## 9. Comparar predicciones y valores reales
**Concepto:** `predict()` indica la especie estimada por el clasificador para cada flor.

```python
y_pred = modelo.predict(X_test)
print("Predicciones:    ", y_pred)
print("Etiquetas reales:", y_test)
print("Predicciones correctas:", int(np.sum(y_pred == y_test)), "de", len(y_test))
```

## 10. Comparación con un modelo básico
**Concepto:** un `DummyClassifier` sirve como referencia sencilla. Ayuda a saber si el clasificador supera una estrategia que predice la clase más frecuente.

```python
base = DummyClassifier(strategy="most_frequent")
base.fit(X_train, y_train)
print("Exactitud del modelo básico: %.2f %%" % (base.score(X_test, y_test) * 100))
print("Exactitud del modelo SVC:    %.2f %%" % (test_results * 100))
```

## 11. Conclusiones

- **Separación 80/20:** permite reservar datos para una evaluación final.
- **Validación cruzada:** ofrece cinco resultados y permite observar la variación de la exactitud.
- **Estandarización:** se realiza dentro de cada entrenamiento para no utilizar información de validación o prueba.
- **SVC:** clasifica flores de Iris según sus cuatro características.
- **Limitación:** el ejercicio usa un conjunto pequeño y los hiperparámetros son fijos; una puntuación alta en esta partición no garantiza idéntico desempeño con datos nuevos.

**Interpretación de las salidas:** compara la media de validación cruzada, la exactitud final en prueba y el clasificador básico. Los resultados numéricos aparecen en las celdas ejecutadas.
