## 1. Importación de librerías

NumPy permite trabajar con matrices y scikit-learn proporciona herramientas de clasificación, preparación de datos y evaluación.

```python
import numpy as np
from sklearn import datasets
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import cross_val_predict, cross_validate, cross_val_score
from sklearn.model_selection import KFold
import sklearn.metrics as metrics
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from pprint import pprint
from sklearn.svm import SVC
```

## 2. Carga de datos

Iris es un conjunto de datos integrado en scikit-learn con 150 muestras, cuatro características y tres clases de flores. No requiere un CSV externo.

```python
# Carga de datos.
datos = datasets.load_iris()
print(np.shape(datos.data))
```

## 3. Variables predictoras y objetivo

X contiene las características de las flores; y contiene las etiquetas de sus especies.

```python
# Extraer la matriz de datos "X" y la variable target "y"
X = datos.data
y = datos.target
```

## 4. División de datos (80/20)

Se reserva el 80 % para entrenamiento y el 20 % para prueba. random_state permite reproducir la partición.

```python
# Partición EXTERNA. Test: hold-out split 80-20%.
X_train, X_test, y_train, y_test = train_test_split(datos.data, datos.target, test_size=0.2, random_state=42)
print('Train dimensions: ', np.shape(X_train))
print('Test dimensions:  ', np.shape(X_test))
# Copia sin estandarizar para la validación cruzada.
X_train_original = X_train.copy()
```

## 5. Estandarización

StandardScaler resta la media y divide por la desviación estándar. Se ajusta únicamente con entrenamiento y se aplica a prueba.

```python
# Estandarización de los datos de entrenamiento y test
standardizer = preprocessing.StandardScaler()
X_train = standardizer.fit_transform(X_train)
X_test = standardizer.transform(X_test)
```

## 6. Modelo SVM

SVC clasifica mediante máquinas de vectores de soporte. El kernel RBF permite fronteras no lineales; C regula la penalización de errores y gamma controla la influencia de las muestras.

```python
# Definimos el algoritmo
alg = SVC(C=0.1, gamma='auto', kernel='rbf')
```

## 7. Validación cruzada

Se generan predicciones fuera de muestra en cinco particiones. El pipeline ajusta el escalador dentro de cada partición para evitar fuga de datos.

```python
# Validación cruzada interna de 5 particiones.
from sklearn.pipeline import make_pipeline
pipeline = make_pipeline(preprocessing.StandardScaler(), alg)
y_pred = cross_val_predict(pipeline, X_train_original, y_train, cv=5)
pprint(y_pred)
```

## 8. Métricas de clasificación

Exactitud: proporción de aciertos. Precisión: aciertos entre las predicciones de cada clase. Sensibilidad: proporción de casos reales detectados. F1: media armónica de precisión y sensibilidad. El promedio macro asigna igual peso a cada clase.

```python
# Presentación de los resultados de la evaluación.
print("Exactitud: %.3f" % (metrics.accuracy_score(y_train, y_pred))) # accuracy
print("Precisión: %.3f" % (metrics.precision_score(y_train, y_pred, average="macro"))) # precision
print("Sensibilidad: %.3f" % (metrics.recall_score(y_train, y_pred, average="macro"))) # sensibilidad
print("F1-score: %.3f" % (metrics.f1_score(y_train, y_pred, average="macro"))) # F-score
```

## 9. Matriz de confusión

Las filas representan clases reales y las columnas clases predichas. La diagonal muestra los aciertos.

```python
# Extraemos la matriz de confusión
print("Matriz de confusión:\n", metrics.confusion_matrix(y_train, y_pred))
```

## 10. Informe de clasificación

Presenta precisión, sensibilidad, F1 y soporte (cantidad de muestras) por clase.

```python
print("Tabla de métricas:\n", metrics.classification_report(y_train, y_pred))
```

## 11. Entrenamiento y evaluación final

Se entrena el modelo con todo el conjunto de entrenamiento y se calcula la exactitud en prueba. Los hiperparámetros son los definidos en el ejercicio; no se realiza una búsqueda de óptimos.

```python
# Entrenamiento definitivo con los hiperparámetros definidos.
model = SVC(C=0.1, gamma='auto', kernel='rbf')
model.fit(X_train, y_train)

# Evaluación con datos de prueba independientes.
test_results = model.score(X_test, y_test)
print('Exactitud en test: ', np.round(test_results*100,4), '%')
```

## 12. Predicciones finales

predict obtiene las clases del conjunto de prueba; el informe y la matriz permiten analizar los aciertos y errores finales.

```python
# Predicciones del conjunto de prueba.
y_pred_test = model.predict(X_test)
print(metrics.classification_report(y_test, y_pred_test))
print("Matriz de confusión:\n", metrics.confusion_matrix(y_test, y_pred_test))
```

## 13. Conclusión

El modelo SVM obtuvo una exactitud del 86,67 % en validación cruzada y del 96,67 % en prueba (29 de 30 muestras correctas). Estos resultados corresponden a esta partición de Iris y no garantizan el mismo rendimiento en otros datos.
