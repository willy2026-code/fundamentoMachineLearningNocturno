### Objetivo
Aprender a reemplazar los datos faltantes de un conjunto de entrenamiento mediante la **media de cada columna**, y aplicar el mismo criterio a nuevos datos de prueba.

> **Nota:** Se mantiene el procedimiento y los valores del código mostrado en las imágenes. Las explicaciones se agregan en celdas Markdown para facilitar el estudio.

## 1. ¿Qué son los valores ausentes?

Un **valor ausente** es un dato que no está disponible en una tabla o conjunto de datos. En NumPy se puede representar mediante `np.nan` (*Not a Number*).

**¿Por qué debemos tratarlos?** Muchos algoritmos de aprendizaje automático no admiten datos faltantes. La **imputación** consiste en sustituirlos por un valor calculado a partir de los datos disponibles.

## 2. Importación de bibliotecas

- **NumPy (`numpy`)**: permite crear y manipular datos numéricos y representar valores faltantes con `np.nan`.
- **`SimpleImputer` de scikit-learn**: reemplaza los valores faltantes siguiendo una estrategia; aquí se usa la **media**.

Si aparece `ModuleNotFoundError`, instala las bibliotecas en el entorno de Anaconda que utiliza JupyterLab.

````python
import numpy as np
from sklearn.impute import SimpleImputer
````

## 3. Datos de entrenamiento

`X_train` es un conjunto de **tres filas y dos columnas**. El valor `np.nan` de la segunda fila, primera columna, indica un dato faltante.

| Fila | Columna 1 | Columna 2 |
|---|---:|---:|
| 1 | 1 | 4 |
| 2 | Ausente | 3 |
| 3 | 7 | 8 |

````python
# Carga de datos
X_train = [[1, 4],
           [np.nan, 3],
           [7, 8]]
````

## 4. Imputación con la estrategia de la media

`SimpleImputer(strategy='mean')` calcula la **media aritmética** de los valores disponibles en cada columna, sin contar los `np.nan`.

Para la primera columna:

\[
	ext{Media} = rac{1+7}{2} = 4
\]

Por tanto, el valor faltante se reemplaza por **4**.

**Funciones clave:**
- `fit(X_train)` **aprende** las medias a partir de los datos de entrenamiento.
- `transform(X_train)` **sustituye** los valores ausentes utilizando las medias aprendidas.
- `fit_transform(X_train)` realiza ambos pasos consecutivamente.

En el ejemplo original se utiliza `fit_transform`.

````python
# Limpieza de datos: imputación valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizamos la estrategia de la media

# Aplicar los métodos "fit" y "transform" para imputar los valores perdidos de X_train
X_train_clean = imputer.fit_transform(X_train)
print(X_train_clean)
````

**Resultado esperado:**

```text
[[1. 4.]
 [4. 3.]
 [7. 8.]]
```

La posición que contenía `np.nan` ahora tiene el valor **4.0**. Los demás datos se conservan; el resultado aparece como números decimales porque NumPy utiliza un arreglo numérico de tipo flotante.

## 5. Imputación de los datos de prueba

`X_test` es un conjunto nuevo que contiene un dato faltante en la **segunda columna**.

**Importante:** se utiliza `imputer.transform(X_test)` y **no** `fit_transform(X_test)`. Así se conservan las medias calculadas con los datos de entrenamiento y se evita aprender nuevos parámetros del conjunto de prueba.

En el entrenamiento, la media de la segunda columna es:

\[
	ext{Media} = rac{4+3+8}{3} = 5
\]

Por eso, el dato ausente en `X_test` se reemplaza por **5**.

````python
# Imputación de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
          [6,     np.nan],
          [8,     2]]

# Imputar los datos del conjunto de test
X_test_clean = imputer.transform(X_test)
print(X_test_clean)
````

**Resultado esperado:**

```text
[[ 4. 10.]
 [ 6.  5.]
 [ 8.  2.]]
```

- El valor faltante de la **primera columna** se reemplaza por **4**, media aprendida del entrenamiento.
- El valor faltante de la **segunda columna** se reemplaza por **5**, media aprendida del entrenamiento.

## 6. Verificación de los resultados

Esta celda adicional comprueba que los valores imputados coinciden con los resultados esperados de las imágenes. Si todo está correcto, aparecerá el mensaje de confirmación.

````python
# Comprobación de los resultados
assert np.array_equal(X_train_clean, np.array([[1., 4.], [4., 3.], [7., 8.]]))
assert np.array_equal(X_test_clean, np.array([[4., 10.], [6., 5.], [8., 2.]]))
print("Verificación correcta: los valores ausentes fueron imputados con la media del entrenamiento.")
````

## 7. Conclusión

La imputación permite preparar conjuntos de datos incompletos para su análisis. En este ejercicio, `SimpleImputer` sustituyó los valores faltantes por la **media de cada columna**, aprendida exclusivamente del conjunto de entrenamiento. El uso posterior de `transform` en los datos de prueba permite aplicar de manera coherente el tratamiento aprendido.

**Recuerda:** la media no siempre es la mejor estrategia; si hay valores extremos, puede convenir la mediana. La elección depende de las características de los datos.
