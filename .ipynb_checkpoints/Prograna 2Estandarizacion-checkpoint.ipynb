## 1. Conceptos clave

- **Dato:** valor numérico que vamos a procesar.
- **Matriz:** conjunto de datos organizado en filas y columnas.
- **Estandarización:** transforma cada columna restando su media y dividiendo entre su desviación estándar.
- **Media:** promedio de los valores de una columna.
- **Desviación estándar:** medida de cuánto se dispersan los valores respecto de la media.
- **`StandardScaler`:** herramienta de `scikit-learn` para estandarizar columnas.

**Fórmula:** $z = \frac{x-\mu}{\sigma}$, donde $x$ es el dato, $\mu$ la media y $\sigma$ la desviación estándar. Tras la transformación, cada columna de estos datos tiene media aproximadamente 0 y desviación estándar aproximadamente 1.

## 2. Importar las librerías

**¿Qué hace?** `numpy` maneja matrices y `preprocessing` contiene `StandardScaler`.



```python
from sklearn import preprocessing
import numpy as np

print("Librerías importadas correctamente")
```

## 3. Cargar los datos originales

**¿Qué hace?** crea `X_train`, una matriz de **3 filas y 3 columnas**. Cada columna se estandariza por separado.

```python
X_train = np.array([[1., -1., 2.],
                    [2.,  0., 0.],
                    [0.,  1., -1.]])
print("Datos originales:")
print(X_train)
print("Dimensión:", X_train.shape)
```

## 4. Crear el estandarizador

**¿Qué hace?** prepara un objeto que aprenderá la **media** y la **desviación estándar** de cada columna.

`StandardScaler()` todavía **no** transforma la matriz; solo crea la herramienta.

```python
standardizer = preprocessing.StandardScaler()
print("Estandarizador creado:", standardizer)
```

## 5. Estandarizar con 'fit_transform()'

**¿Qué hace?** ejecuta dos operaciones:

1. `fit`: calcula media y desviación estándar de `X_train`.
2. `transform`: usa esos valores para estandarizar los datos.

**Resultado esperado:** valores negativos, positivos y ceros; eso es normal.

```python
X_train_std = standardizer.fit_transform(X_train)
print("Matriz estandarizada:")
print(X_train_std)
```

## 6. Verificar las medias y desviaciones

**¿Qué hace?** comprueba que cada columna quedó centrada cerca de 0 y escalada cerca de 1.

**Nota:** pequeñas diferencias numéricas pueden aparecer por redondeo de punto flotante.

```python
print("Media original por columna:", X_train.mean(axis=0))
print("Desviación original por columna:", X_train.std(axis=0))
print("Media estandarizada por columna:", np.round(X_train_std.mean(axis=0), 6))
print("Desviación estandarizada por columna:", np.round(X_train_std.std(axis=0), 6))
```

## 7. Segunda forma: `fit()` y `transform()` por separado

Esta sección reproduce la segunda estrategia del PDF.

- **`fit(X_train)`**: aprende los parámetros únicamente de los datos de entrenamiento.
- **`transform(X_train_2)`**: transforma otra matriz **usando los parámetros ya aprendidos**.

Aquí `X_train_2` tiene exactamente los mismos valores para poder comparar.

```python
X_train_2 = np.array([[1., -1., 2.],
                      [2.,  0., 0.],
                      [0.,  1., -1.]])

f = preprocessing.StandardScaler()
f = f.fit(X_train)
X_train_2_std = f.transform(X_train_2)
print("Resultado con fit() + transform():")
print(X_train_2_std)
```

## 8. Comparar las dos matrices

**¿Qué hace?** `np.array_equiv()` comprueba si los valores y las formas de las matrices son iguales.

**Resultado esperado:** `True`, porque se usaron los mismos datos y parámetros.

```python
print("¿Las matrices son iguales?:", np.array_equiv(X_train_std, X_train_2_std))
assert np.array_equiv(X_train_std, X_train_2_std), "Las matrices no coinciden"
print("✅ Comprobación superada")
```

## 9. Ejercicio adicional: transformar nuevos datos

**Ampliación didáctica:** no aparece en el PDF original.

En un proyecto real, se ajusta el estandarizador con los datos de entrenamiento y se aplica a nuevos registros **sin volver a ejecutar `fit()`**. Así se mantiene la misma escala.

```python
nuevos_datos = np.array([[3., 2., 1.]])
nuevos_estandarizados = standardizer.transform(nuevos_datos)
print("Nuevo registro:", nuevos_datos)
print("Nuevo registro estandarizado:", nuevos_estandarizados)
```
