```python
# Importamos la librerias
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

```python
# Creamos los datos
X = np.array([0.5, 23, 0.3, 4.5, 0.3, 0.5, -28, -50, 60, -100, -10, -11, 13, 19, 1, 9])

print(X)
```

```python
# Calculamos medidas de tendencia central
media = np.mean(X)
mediana = np.median(X)
moda, _ = stats.mode(X)

print("La media de los datos es : ", media)
print("La mediana de los datos es : ", mediana)
print("La moda del conjunto de datos es: ", moda)
```

```python
# Calculamos medidas de posición
Q1 = np.percentile(X, 25)
Q3 = np.percentile(X, 75)
RIC = Q3 - Q1

lim_inf = Q1 - 1.5*RIC
lim_sup = Q3 + 1.5*RIC

print("Limite inferior ", lim_inf)
print("Limite superior ", lim_sup)
print("RIC: ", RIC)
```

```python
#Dibujamos una caja de bigotes para ver los datos
plt.boxplot(X)
plt.grid()
plt.title("Diagrama de caja de bigotes")
plt.show()
```
