# Variables cualitativas
**Estadística descriptiva con Python**

 Se estudia el nivel de estudios, una variable cualitativa ordinal, mediante frecuencias y gráficos. Los datos son simulados.

## 1. Importación de librerías
NumPy permite trabajar con arreglos; random genera números pseudoaleatorios. Se añade una semilla de NumPy para reproducir el orden aleatorio.

```python
import numpy as np
import random
random.seed(42)
np.random.seed(42)
```

## 2. Creación de los datos
`extend()` agrega elementos a la lista. `np.random.permutation()` mezcla los datos sin alterar las cantidades de cada categoría.

```python
nivel_estudios = []
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)

nivel_estudios = np.random.permutation(nivel_estudios)
```

## 3. Visualización de los datos
`print()` muestra el arreglo completo.

```python
print(nivel_estudios)
```

## 4. Frecuencia absoluta
`np.unique()` identifica las categorías y `return_counts=True` cuenta cuántas veces aparece cada una.

```python
import matplotlib.pyplot as plt
import numpy as np
estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

## 5. Diagrama de barras
`plt.bar()` representa las frecuencias mediante barras; los ejes indican categoría y cantidad.

```python
plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras')
plt.bar(estudios, frecuencia)
plt.show()
```

## 6. Diagrama de sectores
`plt.pie()` representa la proporción de cada categoría. `autopct` muestra porcentajes con dos decimales. Se utiliza una lista de colores en lugar del conjunto del PDF para fijar su orden.

```python
colores = ['red', 'blue', 'orange', 'green', 'cyan']
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```

## 7. Tarea resuelta: implementación con pandas
Se solicita implementar la misma información con pandas. `DataFrame` organiza los datos y `value_counts()` calcula las frecuencias. La frecuencia relativa es la cantidad dividida para el total.

```python
import pandas as pd

df = pd.DataFrame({'Nivel de estudios': nivel_estudios})
conteo = df['Nivel de estudios'].value_counts().sort_index()
tabla = conteo.rename('Frecuencia absoluta').to_frame()
tabla['Frecuencia relativa'] = conteo / len(df)
tabla['Porcentaje (%)'] = tabla['Frecuencia relativa'] * 100
display(tabla)
print('Total de datos:', len(df))
assert tabla['Frecuencia absoluta'].sum() == len(df)
```

### Gráficos con pandas
`plot.bar()` y `plot.pie()` permiten representar la tabla de frecuencias directamente.

```python
conteo.plot.bar(title='Diagrama de barras con pandas', rot=0)
plt.xlabel('Nivel de estudios')
plt.ylabel('Frecuencia absoluta')
plt.tight_layout()
plt.show()
```

```python
conteo.plot.pie(autopct='%1.2f%%', title='Diagrama de sectores con pandas', ylabel='')
plt.show()
```

## 8. Interpretación
De los 500 registros, Bachillerato es la categoría más frecuente (172; 34,4 %) y Doctorado la menos frecuente (21; 4,2 %). Las frecuencias no cambian al mezclar los datos. Los gráficos muestran la distribución de la muestra simulada, no de una población real.
