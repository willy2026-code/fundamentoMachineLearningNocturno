**Normalizacion:** Proceso en el cual convertimos nuestros datos de 0 y 1. A
esteproceso se le denomina Homotecia.

**Formula:**
$
x_i^N = \left\{ \frac{x_i - \min(x_i)}{\max(x_i) - \min(x_i)}, \quad \forall j \in \{1,\ldots,n\} \right\}
$

```python
import numpy as np
from sklearn import preprocessing

np.set_printoptions(precision=3, suppress=True)
print("Bibliotecas importadas correctamente")
```

##  Ingresar los datos de entrenamiento

Cada fila representa una observación y cada columna una característica. **Puedes cambiar los números** conservando tres columnas.

```python
X_train = np.array([
    [1., -4.,  5.],
    [5.,  0.,  2.],
    [0.,  6., -3.]
])

print("Datos originales de entrenamiento:")
print(X_train)
```

## 

Normalizar con MinMaxScaler

Para cada columna, el escalado por defecto se calcula como:

$ x_{normalizado} = \dfrac{x - x_{min}}{x_{max} - x_{min}} $

`fit_transform` aprende los valores mínimos y máximos de **entrenamiento** y aplica la transformación.

```python
normalizer = preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)

print("Valores mínimos aprendidos:", normalizer.data_min_)
print("Valores máximos aprendidos:", normalizer.data_max_)
print("\nDatos normalizados de entrenamiento:")
print(X_train_norm)
```

##  Comprobar `fit` y `transform` por separado

El PDF comprueba que llamar por separado a `fit` y `transform` da el mismo resultado que `fit_transform`.

```python
X_train_2 = np.copy(X_train)
f = normalizer.fit(X_train_2)
X_train_norm_2 = f.transform(X_train_2)

print("¿Son iguales ambos resultados?", np.array_equiv(X_train_norm, X_train_norm_2))
assert np.allclose(X_train_norm, X_train_norm_2)
```

##  Transformar los datos de prueba

**Importante:** para los datos de prueba se usa solamente `transform`. No debes volver a ajustar con `fit`, porque se perdería la referencia aprendida en entrenamiento. Los datos de prueba pueden quedar fuera del rango de 0 a 1 si contienen valores fuera de los extremos de entrenamiento.

```python
X_test = np.array([
    [1., -3.,  2.],
    [3.,  0.,  0.],
    [0.,  4., -1.]
])

X_test_norm = f.transform(X_test)
print("Datos de prueba originales:")
print(X_test)
print("\nDatos de prueba normalizados:")
print(X_test_norm)
```

## Verificar los resultados 

Estas comprobaciones corresponden a las matrices incluidas en el documento.

```python
esperado_train = np.array([[0.2, 0., 1.], [1., 0.4, 0.625], [0., 1., 0.]])
esperado_test = np.array([[0.2, 0.1, 0.625], [0.6, 0.4, 0.375], [0., 0.8, 0.25]])
np.testing.assert_allclose(X_train_norm, esperado_train)
np.testing.assert_allclose(X_test_norm, esperado_test)
print("✅ Resultados iguales a los del PDF original")
```

## 7. EJemplo de Práctica editable (independiente)

Modifica las matrices siguientes y vuelve a ejecutar **solo esta celda**. Deben tener el mismo número de columnas. Este ejemplo usa un escalador nuevo para no afectar al ejercicio original.

```python
mis_datos_entrenamiento = np.array([
    [10., 100., 5.],
    [20., 200., 4.],
    [30., 300., 6.]
])
mis_datos_prueba = np.array([
    [15., 150., 3.],
    [25., 250., 5.]
])

mi_escalador = preprocessing.MinMaxScaler()
mis_datos_normalizados = mi_escalador.fit_transform(mis_datos_entrenamiento)
mi_prueba_normalizada = mi_escalador.transform(mis_datos_prueba)

print("Entrenamiento normalizado:\n", mis_datos_normalizados)
print("\nPrueba normalizada:\n", mi_prueba_normalizada)
```
